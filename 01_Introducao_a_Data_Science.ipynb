# Pandas

```python
import pandas as pd
```

```python
notas = pd.read_csv("https://raw.githubusercontent.com/alura-cursos/data-science-analise-exploratoria/main/Aula_0/ml-latest-small/ratings.csv")
```

```python
notas
```

```python

notas.shape # atributo de nota que apresenta um descritivo dessa tabela (número de linhas e colunas)
```

```python
notas.columns = ["usuarioId", "filmeId", "nota", "momento"]
notas
```

```python
notas.head() # função python que traz as 5 primeiras linhas da tabela
```

```python
notas["nota"] # maneira de trazer uma única coluna
```

```python
notas["nota"].unique() #função que apresenta valores únicos
```

```python
notas["nota"].value_counts() # conta quantas vezes cada nota aparece
```

```python
notas["nota"].mean() #apresenta a média de uma coluna
```

```python
notas["nota"].plot(kind='hist', title = "Nº de Notas") # mostra em gráfico a quantidade que as notas aparecem
```

```python
notas["nota"].median()
```

```python
mediana = notas["nota"].median()
media = notas["nota"].mean()
print(f"Mediana é {mediana}")
print(f"Média é {media}")
```

```python
notas["nota"].describe() # da uma panorama geral dos dados da tabela
```

# Seaborn

```python
import seaborn as sns

sns.boxplot(notas["nota"]) # mostra para nós aonde está a mediana da nossa coluna
```

Explorando Dados


```python
filmes = pd.read_csv("https://raw.githubusercontent.com/alura-cursos/data-science-analise-exploratoria/main/Aula_0/ml-latest-small/movies.csv")
filmes.columns = ["filmeId", "titulo", "generos"]
filmes.head()
```

```python
notas.head() # voltei a solicitar as 5 primeiras informações da tabela notas
```

```python
notas.query("filmeId==1") # solicitei checar todas as notas do filme de ID 1
```

```python
notas.query("filmeId==1")["nota"].mean() # solicitei visualizar a média da coluna notas atribuidas ao filme de Id 1
```

```python
notas.query("filmeId==2")["nota"].mean()
```

```python
medias_por_filme = notas.groupby("filmeId")["nota"].mean() # solicitei as medias por filmes, agrupei em um variavel a chamei
medias_por_filme.head()
```

```python
medias_por_filme.plot(kind="hist", title="Nota Média dos Filmes", xlabel="notas médias", ylabel="Frequência") # criei um gráfico do tipo histograma e rotulei titulo, x e y.
```
