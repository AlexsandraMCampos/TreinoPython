```python
# Identificar o dia da semana
# receba um número e verificar o dia da semana. Exemplo 1 = domingo

dia = int(input("Digite um número de 1 a 7: "))

match dia:
    case 1:
        print("Hoje é domingo.")
    case 2:
        print("Hoje é segunda-feira.")
    case 3:
        print("Hoje é terca-feira.")
    case 4:
        print("Hoje é quarta-feira.")
    case 5:
        print("Hoje é quinta-feira.")
    case 6:
        print("Hoje é sexta-feira.")
    case 7:
        print("Hoje é sábado.")
    case _:
        print("Número inválido, digite um número entre 1 e 7.")
```

```python
a = int(input("Digite o valor do lado a: "))
b = int(input("Digite o valor do lado b: "))
c = int(input("Digite o valor do lado b: "))

match a==b, b==c, a==c:# variavéis a serem comparadas
  case (True, True, True):
    print("Os lados são iguais e este é um triângulo equilátero.")
  case (False,False, False):
    print("Os lados nâo são iguais e este é um triângulo escaleno.")
  case _:
    print("Dois lados são iguais e este é um triângulo isosceles.")
```

2. Faça um programa que recebendo um valor inteiro, informe se o número é
positivo, negativo ou neutro.

13. Faça um programa que pergunte ao usuário se ele quer passar uma
temperatura de Fahrenheit para Celsius ou de Celsius para Fahrenheit, e que, a
partir da resposta do usuário, faça a devida conversão.

```python
temperatura = float(input("Informe a temperatura: "))
conversao = input("Você deseja converter essa temperatura para Celsius ou Fahrenheit? :").lower().strip()

match conversao:
    case "celsius":
      temperatura = (temperatura -32)* 5/9
      print(f"A temperatura em °C é {temperatura:.2f}.")
    case "fahrenheit":
      temperatura = (temperatura *9/5)+32
      print(f"A temperatura em °F é {temperatura:.2f}.")
    case _:
      print("Opção inválida! Escolha entre Celsius ou Fahrenheit.")

```

As perguntas são:
“Telefonou para a vítima? “
“Esteve no local do crime?”
“Mora perto da vítima? “
“Devia para a vítima? “
“Já trabalhou com a vítima? “

O programa deve no final emitir uma classificação sobre a participação da
pessoa no crime. Se a pessoa responder positivamente a 2 questões ela deve
ser classificada como “Suspeita”, entre 3 e 4 como “Cúmplice” e 5 como
“Assassino “. Caso contrário, ele será classificado como “Inocente “.

```python
p1= int(input("Telefonou para a vítima. Digite 1 para sim e O para não:"))
p2 = int(input("Esteve no local do crime? Digite 1 para sim e O para não:"))
p3 =int( input("Mora perto da vítima? Digite 1 para sim e O para não:"))
p4 = int(input("Devia para a vítima?  Digite 1 para sim e O para não:"))
p5 =int( input("Já trabalhou com a vítima? Digite 1 para sim e O para não:"))

respostas_positivas= sum([p1,p2,p3,p4,p5])

match respostas_positivas:
  case 5:
    print("Está pessoa é assassina.")
  case 3|4:
    print("Está pessoa é cúmplice.")
  case 2:
    print("Está pessoa é suspeita.")
  case _:
    print("Está pessoa é inocente.")
```
